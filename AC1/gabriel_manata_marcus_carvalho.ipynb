```python
from matplotlib import pyplot as plt
import numpy as np
import math
```

```python
derivadas = [lambda x: -np.sin(x), lambda x: -np.cos(x), lambda x: np.sin(x), lambda x: np.cos(x)]
def taylor(quant_termos, x0, h):
    fx0 = np.cos(x0)  
    for n in range(quant_termos):
        fx0 += derivadas[n % len(derivadas)](x0) * (h ** (n + 1)) / math.factorial(n + 1)
    return round(float(fx0),6)
```

```python
f = lambda x: np.cos(x)

quant_termos = int(input("Digite a quantidade de termos desejados: "))
x0  = float(input("Digite o termo inicial: "))
x = float(input("Digite o termo final: "))
h = x - x0
eixo_x = np.linspace(x0-1, x+1 ,100)
plt.plot(eixo_x, f(eixo_x), "b", label="f(x) = cos(x)")
plt.plot(x0, f(x0), "r.", label="Ponto Inicial")
plt.plot(x, f(x), "bd", label="Ponto Final")
plt.title("Cenário Inicial")
plt.legend()
plt.grid()
plt.show()
```

```python
t = taylor(quant_termos, x0, h)
t
```

```python
erro = abs(round(float(((f(x) - t) / f(x)) * 100), 2))
erro
```

```python
print(f"O valor da função (y) no ponto x0 = {x0} é {f(x0)}")
print(f"O valor da função (y) no ponto x = {x} é {f(x)}")
print(f"O valor aproximado de Taylor na ordem {quant_termos} é {t}")
print(f"O erro relativo a aproximação de ordem {quant_termos} é de {erro}%")

plt.plot(eixo_x, f(eixo_x), "b", label="f(x) = cos(x)")
plt.plot(x0, f(x0), "r.", label="Ponto Inicial")
plt.plot(x, f(x), "bd", label="Ponto Final")
plt.plot(x, t, "bx", label=f"Taylor de ordem {quant_termos}")
plt.legend()
plt.grid()
plt.show()
```
