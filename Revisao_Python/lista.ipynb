## QUESTÃO 2

```python
l1 = int(input("Digite o lado 1 do triângulo: "))
l2 = int(input("Digite o lado 2 do triângulo: "))
l3 = int(input("Digite o lado 3 do triângulo: "))

lados = [l1, l2, l3]
lados.sort()

if lados[0] + lados[1] <= lados[2]:
    print("Não é um triângulo válido")
else:
    print("Triângulo Válido!")

print([f"Lado: {l}" for l in lados])

if l1 == l2 and l1 == l3:
    print("Triângulo Equilátero")
elif l1 == l2 or l1 == l3 or l2 == l3:
    print("Triângulo Isósceles")
else:
    print("Triângulo Escaleno")
```

## QUESTÃO 3

```python
valores = []
while True:
    v = int(input("Digite um número inteiro positivo: "))
    if v <= 0:
        break
    valores.append(v)
print(f"Valores apresentados: {valores}")
print(f"Maior valor: {max(valores)}")
print(f"Menor valor: {min(valores)}")
print(f"Quantidade de valores: {len(valores)}")
print(f"Soma de valores: {sum(valores)}")
print(f"Média de valores: {sum(valores)/ len(valores)}")
```

## QUESTÃO 5

```python
def maior(lista):
    lista.sort()
    return lista[-1]
def menor(lista):
    lista.sort()
    return lista[0]
def soma(lista):
    soma = 0
    for _ in lista:
        soma += _
    return soma
def media(lista):
    soma = 0
    for _ in lista:
        soma += _
    media = soma / len(lista)
    return media
    
```
