```python
from matplotlib import pyplot as plt
import numpy as np
```

$$
x^3 - 0.75x^2 + 3 
$$

```python
funcao = lambda x: x**3 - 0.75*x**2 + 3 
```

## Plotar o gráfico inicial

```python
# Indo de 0 a 1
x0 = 0.5
x = 1
# plot(eixo x, eixo y, forma geometrica (ro - red Oval - para ponto e bd- blue diamond - para diamante), label)
plt.plot(x0, funcao(x0), 'ro', label='ponto inicial') # 0.5 na função deu ~2.93.
plt.plot(x, funcao(x), 'bd', label='ponto final')
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("F(x)")
plt.show()
```

## Calcular termo de Primeira ordem
$$
t(1) \approx f(0.5) + \frac{f'(0.5)*(1 - 0.5)}{1!} \approx 2.9375
$$

```python
derivada_1 = lambda x: 3*x**2 - 1.5*x

t1 = funcao(x0) + (derivada_1(x0) * (x - x0)) / 1

t1
```

## Plotar com termo de Primeira ordem

Com os elementos de taylor eu quero sempre me APROXIMAR DO X FINAL, por isso eu ploto ele com o mesmo valor no eixo X e vou variando no eixo Y.

```python
plt.plot(x0, funcao(x0), "ro", label="Ponto Inicial")
plt.plot(x, funcao(x), "bd", label="Ponto Final")
plt.plot(x, t1, "gd", label="Taylor ordem 1") 
plt.grid()
plt.xlabel("Eixo X")
plt.ylabel("F(x)")
plt.legend()
plt.show()

print(f"Taylor 1 deu um erro de {((funcao(x) - t1)/funcao(x)) * 100:.4f}%")
```

## Continuar calculando os taylors até chegar próximo o suficiente

Taylor de segunda:
$$
t(2) \approx f(0.5) + \frac{f'(0.5) * (1 - 0.5)}{1!} + \frac{f''(0.5) * (1 - 0.5)^2}{2!} \approx 3.125
$$

```python
derivada_2 = lambda x: 6*x - 1.5
t2 = t1 + (derivada_2(x0) * (x - x0)**2)/ 2
erro_2 = round(((funcao(x) - t2)/ funcao(x)) *100, 4)

plt.plot(x0, funcao(x0), "ro", label="Ponto Inicial")
plt.plot(x, funcao(x), "bd", label="Ponto Final")
plt.plot(x, t1, "gd", label="Taylor ordem 1") 
plt.plot(x, t2, "rs", label="Taylor ordem 2")
plt.grid()
plt.xlabel("Eixo X")
plt.ylabel("F(x)")
plt.legend()
plt.show()
print(f"Taylor 2 deu um erro de {erro_2}%")
```

Taylor de terceira: 
$$
t(3) \approx f(0.5) + \frac{f'(0.5) * (1 - 0.5)}{1!} + \frac{f''(0.5) * (1 - 0.5)^2}{2!}  + \frac{f'''(0.5) * (1 - 0.5)^3}{3!} \approx 3.25
$$

No Taylor de terceira o valor deu exato o valor de X final, temos erro de 0%

```python
derivada_3 = 6
t3 = t2 + (6 * ((1 - 0.5)**3) / 6)
erro_3 = round(((funcao(x) - t3) / funcao(x)) *100,4)


plt.plot(x0, funcao(x0), "ro", label="Ponto Inicial")
plt.plot(x, funcao(x), "bd", label="Ponto Final")
plt.plot(x, t1, "gd", label="Taylor ordem 1") 
plt.plot(x, t2, "rs", label="Taylor ordem 2")
plt.plot(x, t3, "gs", label="Taylor de terceira")
plt.grid()
plt.xlabel("Eixo X")
plt.ylabel("F(x)")
plt.legend()
plt.show()
print(f"Taylor 3 deu um erro de {erro_3}%")
```
