Questão 1. Use a série de Taylor para fazer uma estimativa $f(x) = e^{-x}$ em $x_{i+1} = 1$ para $x_i = 0{,}2$. Use as versões de ordem zero, um, dois e três e calcule $|\varepsilon_t|$ para cada caso.


```python
from matplotlib import pyplot as plt
import numpy as np
import sympy as sp
```

```python
F = lambda x: np.e**-x
x0 = 0.2
xi = 1
h = xi - x0
```

```python
x = sp.symbols('x')
f_simul = sp.exp(-x)
d_f = sp.diff(f_simul,x)
d_f2 = sp.diff(d_f, x)
d_f3 = sp.diff(d_f2,x)
print(d_f)
print(d_f2)
print(d_f3)
```

```python
d_f = lambda x: -np.e**-x
d_f2 = lambda x: np.e**-x
d_f3 = lambda x: -np.e**-x
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

```python
t0 = F(0.2)
round(t0,4)
```

```python
e0 = ((F(1) - t0) /  F(1)) * 100
e0
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.plot(xi, t0, "bs", label="Taylor 0")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()

print(f"Erro de Taylor 0 é de {round(-1 *e0, 4)}%")
```

```python
t1 = F(x0) + (d_f(x0) * h)
e1 = round(((F(xi) - t1) / F(xi) * 100) , 4)
e1
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.plot(xi, t0, "sb", label="Taylor 0")
plt.plot(xi, t1, "rs", label="Taylor 1")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
print(f"Erro de Taylor 1 é de {round(e1, 4)}%")
```

```python
t2 = t1 + (d_f2(x0) * h**2)/2
t2
```

```python
e2 = ((F(xi) - t2)/ F(xi)) * -100
e2
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.plot(xi, t0, "bs", label="Taylor 0")
plt.plot(xi, t1, "rs", label="Taylor 1")
plt.plot(xi, t2, "gs", label="Taylor 2")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
print(f"Erro de Taylor 2 é de {round(e2, 4)}%")
```

```python
t3 = t2 + (d_f3(x0) * h**3)/6
t3
```

```python
e3 = ((F(xi) - t3)/F(xi)) * 100 
e3
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.plot(xi, t0, "bs", label="Taylor 0")
plt.plot(xi, t1, "rs", label="Taylor 1")
plt.plot(xi, t2, "gs", label="Taylor 2")
plt.plot(xi, t3, "k.", label="Taylor 3")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
print(f"Erro de Taylor 2 é de {round(e3, 4)}%")
```

Questão 2. Use a expansão em série de Taylor de ordem zero até três para prever $f(3)$ para

$$
f(x) = 25x^{3} - 6x^{2} + 7x - 88
$$

utilizando o ponto-base $x=1$. Calcule o erro relativo percentual $\varepsilon_t$ para cada aproximação.


```python
import numpy as np
from matplotlib import pyplot as plt
import sympy as sp
```

```python
f = lambda x: 25*x**3 - 6*x**2 + 7*x - 88
x = 3
x0 = 1
h = x - x0
```

```python
x_ficticio = sp.symbols("x")
f_ficticio = 25*x_ficticio**3 - 6*x_ficticio**2 + 7*x_ficticio - 88
df = sp.diff(f_ficticio, x_ficticio)
dff = sp.diff(df, x_ficticio)
dfff = sp.diff(dff, x_ficticio)
print(f"Derivada de Primeira: {df}")
print(f"Derivada de Segunda: {dff}")
print(f"Derivada de Terceira: {dfff}")
```

```python
df = lambda x: 75*x**2 - 12*x + 7
dff = lambda x: 150*x - 12
dfff = 150
```

```python
plt.plot(x0, f(x0), "dr", label="Ponto Inicial")
plt.plot(x, f(x), "dg", label="Ponto Final")
plt.grid()
plt.legend()
plt.show()
```

```python
t0 = f(x0)
e0 = ((f(x) - t0) / f(x)) * 100
print(t0, e0)
```

```python
plt.plot(x0, f(x0), "dr", label="Ponto Inicial")
plt.plot(x, f(x), "dg", label="Ponto Final")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.grid()
plt.legend()
plt.show()
print(f"Taylor 0 com erro de {round(e0,4)}%")
```

```python
t1 = t0 + (df(x0) * h)
e1 = ((f(x) - t1) / f(x)) * 100
print(t1, e1)
```

```python
plt.plot(x0, f(x0), "dr", label="Ponto Inicial")
plt.plot(x, f(x), "dg", label="Ponto Final")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 0")
plt.grid()
plt.legend()
plt.show()
print(f"Taylor 1 com erro de {round(e1,4)}%")
```

```python
t2 = t1 + (dff(x0) * h**2)/2
e2 = ((f(x) - t2)/f(x)) * 100
print(t2, e2)
```

```python
plt.plot(x0, f(x0), "dr", label="Ponto Inicial")
plt.plot(x, f(x), "dg", label="Ponto Final")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 1")
plt.plot(x, t2, "m.", label="Taylor 2")
plt.grid()
plt.legend()
plt.show()
print(f"Taylor 2 com erro de {round(e2,4)}%")
```

```python
t3 = t2 + (dfff * h**3)/6
e3 = ((f(x) - t3) / f(x)) * 100
print(t3, e3)
```

```python
plt.plot(x0, f(x0), "dr", label="Ponto Inicial")
plt.plot(x, f(x), "dg", label="Ponto Final")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 1")
plt.plot(x, t2, "y.", label="Taylor 2")
plt.plot(x, t3, "c.", label="Taylor 3")
plt.grid()
plt.legend()
plt.show()
print(f"Taylor 3 com erro de {round(e3,4)}%")
```

Questão 3. Use a expansão em série de Taylor de ordem zero até quatro para prever $f(2{,}5)$ para $f(x) = \ln x$ utilizando o ponto-base $x =1$. Calcule o erro relativo percentual para cada aproximação. Discuta o significado dos resultados.


```python
from matplotlib import pyplot as plt
import numpy as np
import sympy as sp
```

```python
f = lambda x: np.log(x)
x = 2.5
x0 = 1
h = x - x0
```

```python
x_ficticio = sp.symbols("x")
f_ficticio = sp.log(x_ficticio)

df = sp.diff(f_ficticio, x_ficticio)
dff = sp.diff(df, x_ficticio)
dfff = sp.diff(dff, x_ficticio)
dffff = sp.diff(dfff, x_ficticio)

print(f"f' = {df}")
print(f"f'' = {dff}")
print(f"f''' = {dfff}")
print(f"f'''' = {dffff}")
```

```python
df = lambda x: 1/x
dff = lambda x: -1/x**2
dfff = lambda x: 2/x**3
dffff = lambda x: -6/x**4
```

```python
plt.plot(x, f(x), "gd", label="Ponto Final")
plt.plot(x0, f(x0), "rd", label="Ponto Inicial")
plt.grid()
plt.legend()
plt.show()
```

```python
t0 = f(x0)
e0 = ((f(x) - t0)/f(x)) * 100
print(t0, e0)
```

```python
plt.plot(x, f(x), "gd", label="Ponto Final")
plt.plot(x0, f(x0), "rd", label="Ponto Inicial")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.grid()
plt.legend()
plt.show()

print(f"Erro do Taylor 0 é {round(e0,4)}")
```

```python
t1 = t0 + (df(x0) * h)
e1 = abs(((f(x) - t1) / f(x)) * 100)

plt.plot(x, f(x), "gd", label="Ponto Final")
plt.plot(x0, f(x0), "rd", label="Ponto Inicial")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 1")
plt.grid()
plt.legend()
plt.show()

print(f"Taylor 1 é {round(t1, 4)}")
print(f"Erro do Taylor 1 é {round(e1,4)}%")
```

```python
t2 = t1 + (dff(x0) * h**2) / 2
e2 = abs(((f(x) - t2) / f(x)) * 100)

plt.plot(x, f(x), "gd", label="Ponto Final")
plt.plot(x0, f(x0), "rd", label="Ponto Inicial")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 1")
plt.plot(x, t2, "y.", label="Taylor 2")
plt.grid()
plt.legend()
plt.show()

print(f"Taylor 2 é {round(t2, 4)}")
print(f"Erro do Taylor 2 é {round(e2,4)}%")
```

```python
t3 = t2 + (dfff(x0) * h**3) / 6
e3 = abs(((f(x) - t3) / f(x)) * 100)

plt.plot(x, f(x), "gd", label="Ponto Final")
plt.plot(x0, f(x0), "rd", label="Ponto Inicial")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 1")
plt.plot(x, t2, "y.", label="Taylor 2")
plt.plot(x, t3, "c.", label="Taylor 3")
plt.grid()
plt.legend()
plt.show()

print(f"Taylor 3 é {round(t3, 4)}")
print(f"Erro do Taylor 3 é {round(e3,4)}%")
```

```python
t4 = t3 + (dffff(x0) * h**4) / 24
e4 = abs(((f(x) - t4) / f(x)) * 100)

plt.plot(x, f(x), "gd", label="Ponto Final")
plt.plot(x0, f(x0), "rd", label="Ponto Inicial")
plt.plot(x, t0, "k.", label="Taylor 0")
plt.plot(x, t1, "r.", label="Taylor 1")
plt.plot(x, t2, "y.", label="Taylor 2")
plt.plot(x, t3, "c.", label="Taylor 3")
plt.plot(x, t4, "m.", label="Taylor 4")
plt.grid()
plt.legend()
plt.show()

print(f"Taylor 4 é {round(t4, 4)}")
print(f"Erro do Taylor 4 é {round(e4,4)}%")
```
