## Questão 1

Use a **série de Taylor** para fazer uma estimativa de $f(x) = e^{-x}$ em $x_{i+1} = 1$ para $x_i = 0{,}2$. Use as versões de **ordem zero, um, dois e três** e calcule $|\varepsilon_t|$ para cada caso.

$$
\boxed{\,f(x) = e^{-x}\,}
$$

**Dados do problema:**

- **Ponto-base:** $x_i = 0{,}2$
- **Ponto de previsão:** $x_{i+1} = 1$
- **Passo:** $h = x_{i+1} - x_i = 1 - 0{,}2 = 0{,}8$
- **Ordens pedidas:** $0, 1, 2, 3$
- **Erro verdadeiro percentual:** $\varepsilon_t = \dfrac{\text{valor verdadeiro} - \text{valor aproximado}}{\text{valor verdadeiro}} \times 100$

---


```python
from matplotlib import pyplot as plt
import numpy as np
import sympy as sp
```

```python
F = lambda x: np.e**-x
x0 = 0.2
xi = 1
h = xi - x0
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

```python
x = sp.symbols('x')
f_simul = sp.exp(-x)
d_f = sp.diff(f_simul,x)
d_f2 = sp.diff(d_f, x)
d_f3 = sp.diff(d_f2,x)
print(d_f)
print(d_f2)
print(d_f3)
```

```python
d_f = lambda x: -np.e**-x
d_f2 = lambda x: np.e**-x
d_f3 = lambda x: -np.e**-x
```

```python
t1 = round(F(x0) + (d_f(x0) * h)/1, 4)
e1 = round(((F(xi) - t1) / F(xi) * 100) , 4)
e1
```

```python
plt.plot(x0, F(x0), "bd", label="Ponto inicial")
plt.plot(xi, F(xi), "gd", label="Ponto final")
plt.plot(xi, t1, "rs", label="Taylor 1")
plt.grid()
plt.legend()
plt.xlabel("Eixo X")
plt.ylabel("Eixo Y")
plt.show()
```

# Respostas

## Questão 1 — $f(x) = e^{-x}$, $x_i = 0{,}2 \to x_{i+1} = 1$ $(h = 0{,}8)$

Valor verdadeiro: $f(1) = e^{-1} = 0{,}367879$

| Ordem | Taylor $f(1)$ | $\vert\varepsilon_t\vert$ |
|------:|--------------:|------------------:|
| 0 | $0{,}818731$ | $122{,}5541\%$ |
| 1 | $0{,}163746$ | $55{,}4892\%$ |
| 2 | $0{,}425740$ | $15{,}7281\%$ |
| 3 | $0{,}355875$ | $3{,}2632\%$ |

## Questão 2 — $f(x) = 25x^{3} - 6x^{2} + 7x - 88$, $x = 1 \to f(3)$ $(h = 2)$

Valor verdadeiro: $f(3) = 554$

| Ordem | Taylor $f(3)$ | $\varepsilon_t$ |
|------:|--------------:|----------------:|
| 0 | $-62$ | $111{,}1913\%$ |
| 1 | $78$ | $85{,}9206\%$ |
| 2 | $354$ | $36{,}1011\%$ |
| 3 | $554$ | $0{,}0000\%$ |

## Questão 3 — $f(x) = \ln x$, $x = 1 \to f(2{,}5)$ $(h = 1{,}5)$

Valor verdadeiro: $f(2{,}5) = \ln(2{,}5) = 0{,}916291$

| Ordem | Taylor $f(2{,}5)$ | $\varepsilon_t$ |
|------:|------------------:|----------------:|
| 0 | $0$ | $100{,}0000\%$ |
| 1 | $1{,}5$ | $63{,}7035\%$ |
| 2 | $0{,}375$ | $59{,}0741\%$ |
| 3 | $1{,}5$ | $63{,}7035\%$ |
| 4 | $0{,}234375$ | $74{,}4213\%$ |

> **Discussão:** com $h = 1{,}5 > 1$, a série de $\ln x$ fica fora do raio de convergência — o erro não diminui com a ordem.
