```python
import numpy as np
from matplotlib import pyplot as plt
import sympy as sp
```

```python
tempo = np.array([0,1,2,3])
volume = np.array([0,1,2.3,3.9])
plt.plot(tempo, volume, label="Volume", color="g")
plt.xlabel('Tempo(s)')
plt.ylabel('Volume(m3)')
plt.grid()
plt.legend()
plt.show()
```

```python
derivadaordeml=np.gradient(volume)
plt.plot(tempo, volume, label="Volume", color="g")
plt.plot(tempo,derivadaordeml,'ro',label="vazao(m3)")
plt.xlabel('Tempo(s)')
plt.ylabel('Volume(m3)')
plt.grid()
plt.legend()
plt.show()
print(derivadaordeml)
```
