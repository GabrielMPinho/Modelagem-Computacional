## Questao 1

Considere a funcao $f(x) = x^3 - 2x + 4$ no intervalo $[-2, 2]$ com $h = 0{,}25$. 
Use aproximacoes por diferencas finitas progressivas, regressivas e centradas 
para a primeira e segunda derivadas de modo a ilustrar graficamente qual 
aproximacao e mais acurada. Faca o grafico de todas as tres aproximacoes por 
diferencas finitas da primeira derivada com a derivada teorica e faca o mesmo 
tambem para a segunda derivada.

```python
F = lambda x: x**3 - 2*x + 4
h = 0.25
x0 = -2
x = 2
eixox = np.linspace(x0, x, 1000)
pontos_funcao = np.arange(x0, x+0.1, h)

plt.plot(x0, F(x0), "rs", label="Ponto inicial")
plt.plot(x, F(x), "gs", label="Ponto final")
plt.plot(eixox, F(eixox), "r", label="Linspace")
plt.plot(pontos_funcao, F(pontos_funcao), "b.", label="Pontos da função")
plt.grid()
plt.legend()
plt.show()
pontos_funcao
```

## Função progressiva
$$
\frac{f(xi + 1) - f(xi)}{h}
$$

```python
def progressiva_o1(pontos, h):
    valores = []
    for indice in range(len(pontos) - 1): # Como a progressiva pega um valor a frente, quando chegar no ultimo item não vai ter como calcular, ja que não tem item na frente, então o for vai no tamanho do vetor pontos - 1
        valor = (F(pontos[indice+1]) - F(pontos[indice]))/h
        valores.append(valor)
    return valores

val_progressiva = progressiva_o1(pontos_funcao, h)

plt.plot(x0, F(x0), "rs", label="Ponto inicial")
plt.plot(x, F(x), "gs", label="Ponto final")
plt.plot(eixox, F(eixox), "r", label="Linspace")
plt.plot(pontos_funcao[0:16] ,val_progressiva, "-go", label="Valores da progressiva de primeira ordem" )
plt.grid()
plt.legend()
plt.show()
val_progressiva
```

## Função regressiva

$$
\frac{f(xi) - f(xi-1)}{h}
$$

```python
def regressiva_o1(pontos, h):
    valores = []
    for indice in range(1, len(pontos)):
        valor = (F(pontos[indice]) - F(pontos[indice - 1]))/h
        valores.append(valor)
    return valores

val_regressiva = regressiva_o1(pontos_funcao, h)
plt.plot(x0, F(x0), "rs", label="Ponto inicial")
plt.plot(x, F(x), "gs", label="Ponto final")
plt.plot(eixox, F(eixox), "r", label="Linspace")
plt.plot(pontos_funcao[0:16] ,val_progressiva, "-go", label="Valores da progressiva de primeira ordem" )
plt.plot(pontos_funcao[1:17] ,val_regressiva, "-bo", label="Valores da progressiva de segunda ordem" )
plt.grid()
plt.legend()
plt.show()
```

## Função centrada

$$
\frac{f(xi + 1) - f(xi-1)}{2 * h}
$$

```python
def centrada_o1(pontos, h):
    valores = []
    for indice in range(1, len(pontos) - 1):
        valor = (F(pontos[indice + 1]) - F(pontos[indice - 1]))/(2*h)
        valores.append(valor)
    return valores

val_centrada = centrada_o1(pontos_funcao, h)
plt.plot(x0, F(x0), "rs", label="Ponto inicial")
plt.plot(x, F(x), "gs", label="Ponto final")
plt.plot(eixox, F(eixox), "r", label="Linspace")
plt.grid()
plt.legend()
plt.show()

plt.figure(figsize=(12,6))
plt.plot(pontos_funcao[0:16] ,val_progressiva, "-go", label="Valores da progressiva de primeira ordem" )
plt.plot(pontos_funcao[1:17] ,val_regressiva, "-bo", label="Valores da regressiva de primeira ordem" )
plt.plot(pontos_funcao[1:16] ,val_centrada, "-ko", label="Valores da centrada de primeira ordem" )
plt.grid()
plt.legend()
plt.show()
```

```python
x = sp.symbols("x")
f = x**3 - 2*x + 4
fd = sp.diff(f,x)
fd = sp.lambdify(x, fd, "numpy") 

plt.figure(figsize=(12,6))
plt.plot(pontos_funcao, fd(pontos_funcao), "-go", label="Função Derivada")
plt.plot(pontos_funcao[0:16] ,val_progressiva, "-ro", label="Valores da progressiva de primeira ordem" )
plt.plot(pontos_funcao[1:17] ,val_regressiva, "-bo", label="Valores da regressiva de primeira ordem" )
plt.plot(pontos_funcao[1:16] ,val_centrada, "-ko", label="Valores da centrada de primeira ordem" )
plt.grid()
plt.legend()
plt.show()
```

## Função centrada de segunda ordem

$$
\dfrac{f(x_{i+1}) - 2\,f(x_i) + f(x_{i-1})}{h^2}
$$

```python
def centrada_o2(pontos, h):
    valores = []
    for i in range(1, len(pontos) - 1):
        resul = (F(pontos[i + 1]) - 2*F(pontos[i]) + F(pontos[i-1]))/h**2
        valores.append(resul)
    return valores

fdd = sp.diff(f,x)
fdd = sp.diff(fdd,x)
fdd = sp.lambdify(x,fdd,"numpy")

valor_centrada_o2 = centrada_o2(pontos_funcao, h)

plt.figure(figsize=(12,6))
plt.plot(pontos_funcao, fdd(pontos_funcao), "-go", label="Função Derivada de segunda ordem")
plt.plot(pontos_funcao[1:16], valor_centrada_o2, "-ko", label="Valores da centrada de segunda ordem")
plt.grid()
plt.legend()
plt.show()
```

## Questao 2

A posicao de um foguete ao longo do tempo e dada pela tabela abaixo. 
Estime a velocidade e a aceleracao do foguete em cada ponto medido 
utilizando diferencas finitas.

| $t$ (s) | 0 | 25 | 50 | 75 | 100 | 125 |
|---|---|---|---|---|---|---|
| $y$ (km) | 0 | 32 | 58 | 78 | 92 | 100 |

```python
tempo = [0,25,50,75,100,125]
y = [0,32,58,78,92,100]


plt.title("Função da posição por tempo")
plt.plot(tempo, y, '-r')
plt.xlabel("Tempo (s)")
plt.ylabel("Y (km)")
plt.grid()
plt.show()

vel = np.gradient(y, tempo)
plt.title("Função da velocidade por tempo")
plt.plot(tempo, vel, '--b', label="Função de velocidade")
plt.xlabel("Tempo (s)")
plt.ylabel("Velocidade (km/s)")
plt.grid()
plt.legend()
plt.show()

acl = np.gradient(vel, tempo)
plt.plot(tempo, acl, '--g')
plt.title("Função da aceleração por tempo")
plt.xlabel("Tempo (s)")
plt.ylabel("Aceleração (km/s^2)")
plt.grid()
plt.show()
```

### Questao 1

Derivadas teoricas: $f'(x) = 3x^2 - 2$ e $f''(x) = 6x$.

**1a derivada** (pontos selecionados, $h = 0{,}25$):

| $-1{,}00$ | $1{,}0000$ | $0{,}3125$ | $1{,}8125$ | $1{,}0625$ |
| $0{,}00$ | $-2{,}0000$ | $-1{,}9375$ | $-1{,}9375$ | $-1{,}9375$ |
| $1{,}00$ | $1{,}0000$ | $1{,}8125$ | $0{,}3125$ | $1{,}0625$ |
| $2{,}00$ | $10{,}0000$ | --- | $8{,}5625$ | --- |

**2a derivada** (pontos internos, $h = 0{,}25$):

| $x$ | $f''(x)$ teorica | Centrada |
|---|---|---|
| $-1{,}75$ | $-10{,}5000$ | $-10{,}5000$ |

| $1{,}00$ | $6{,}0000$ | $6{,}0000$ |

> A diferenca centrada e mais acurada para a 1a derivada (erro $O(h^2)$) e coincide exatamente com a teorica para a 2a derivada desta funcao cubica (pois $f'''(x) = 6$ e $f''''(x) = 0$, anulando os termos de erro).

Valor verdadeiro: $f'(2) = 75(2)^2 - 12(2) + 7 = 300 - 24 + 7 = 283$.

| Metodo | Aproximacao | Erro $\varepsilon_t$ (%) |
